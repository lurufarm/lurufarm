# 예외 처리
## 예외 발생

```python
try:
    4 / 0
except:
    print("예외 발생")
```

```python
try:
    a = [1,2]
    print (a[3])
    4 / 0
except IndexError:
    print("인덱싱 할 수 없습니다.")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
```

```python
try:
    4/0
except ZeroDivisionError as e:
    print(e)
```

```python
try:
    f = open('C:\\dev\\frog_song.txt', 'r')
finally:
    print("무조건 출력") # 에러가 발생하든 안하든 finally 구문은 동작함
    f.close()
```

```python
try:
    age = int(input('나이를 입력하세요: '))
except:
    print('입력이 정확하지 않습니다.')
else: # 예외가 발생하지 않으면
    if age <= 19:
        print('미성년자는 출입금지입니다.')
    else:
        print('환영합니다.')
```

```python
try:
    f = open("없는파일", 'r')
except FileNotFoundError:
    pass # FileNotFoundError가 일어나면, 그냥 넘어가겠다는 의미
```

```python
class MyError(Exception):
    pass
def say_nick(nick):
    if nick == '돼지':
        raise MyError() # 닉네임이 돼지면 이 에러를 발생시키겠다는 의미
    print(nick)

try:
    say_nick('곰')
except MyError:
    print('허용되지 않는 별명입니다.')
```
