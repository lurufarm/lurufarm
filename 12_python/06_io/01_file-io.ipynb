# 파일 입출력
## 파일 열기 open()
1. r: 읽기 모드
2. w: 쓰기 모드 (파일 내용을 덮어씀)
3. a: 수정 모드 (파일 끝에 내용이 추가됨)


```python
f = open('C:\\dev\\frog_song.txt', 'r')
# data = f.readline()

while True:
    line = f.readline()
    if not line:
        break
    print(line)

f.close()

# print(data)
```

```python
f = open('C:\\dev\\frog_song.txt', 'r')

data = f.read() # readline은 한 줄씩, read는 다 읽어옴

print(data)
f.close()
```

```python
f = open('C:\\dev\\frog_song.txt', 'a') # 수정 모드
f.write("노래 끝")
f.close()
```

```python
f1 = open('C:\\dev\\frog_song.txt', 'r')

data = f1.read()
print(data)

f1.close()
```

```python
f = open('C:\\dev\\frog_song.txt', 'w') # 쓰기 모드
f.write("노래 끝")
f.close()

f1 = open('C:\\dev\\frog_song.txt', 'r')

data = f1.read()
print(data)

f1.close()
```

```python
with open('C:\\dev\\frog_song.txt', 'r') as f: # with를 사용하면 파일이 자동으로 닫힘
    print(f.readline())
```
