# 시퀀스
## 시퀀스 자료형이란
- string : 문자열 값을 가지는 자료형
- list : 여러 값을 담을 수 있는 가변형 자료형
- tuple : 여러 값을 담을 수 있는 불변형 자료형

```python
# 문자열: 
address = '대한민국 서울시 종로구'
print(address[5])

# 리스트
location = ['서울특별시', '부산광역시', '인천광역시', '광주광역시', '울산광역시']
print(location[1])

# 튜플
nation = ['대한민국', '미국', '대만', '프랑스']
print(nation[0])
```

- 인덱스를 음수로 지정하면 문자열 끝에서부터 요소를 찾는다.

```python
address = '대한민국 서울시 서초구'
print(address[-2])

location = ['서울특별시', '부산광역시', '인천광역시', '광주광역시', '울산광역시']
print(location[-1])

nation = ('대한민국', '미국', '대만', '프랑스')
print(nation[-2])
```

- 인덱스를 사용하여 시퀀스 자료에서 원하는 만큼의 문자열을 추출할 수 있다.

```python
address = '대한민국 서울시 서초구'
print(address[5:8])
print(address[1:12:4]) # 시작 인덱스 : 끝 인덱스 : 오프셋
print(address[9:]) # 9번 인덱스부터 끝까지
print(address[:2]) # 처음부터 2번 인덱스 전까지
print(address[::-1]) # 문자열 뒤집기

location = ['서울특별시', '부산광역시', '인천광역시', '광주광역시', '울산광역시']
print(location[2:4])

nation = ('대한민국', '미국', '대만', '프랑스')
print(nation[0:3])
```

## 시퀀스 자료형의 연산 및 내장 함수
### in

```python
# 값 in 시퀀스 자료형
print('대한민국' in address)
print('대한민국서울시' in address)

location = ['서울특별시', '부산광역시', '인천광역시', '광주광역시', '울산광역시']
print('광주광역시' in location)
print(['서울특별시', '부산광역시'] in location) # False를 반환함. []를 요소 자체로 판단하기 때문에

nation = ('대한민국', '미국', '대만', '프랑스')
print('미국' in nation)
print(('대한민국', '미국') in nation) # 마찬가지로 ()를 요소 자체로 판단하기 때문에 False 출력

print(len(address))     # 문자열의 길이
print(len(location))    # 리스트 요소 갯수
print(len(nation))      # 튜플 요소 갯수
```
