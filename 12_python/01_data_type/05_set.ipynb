# 셋(Set)
## 셋이란
- 중복된 요소를 허용하지 않으며, 순서 없이 요소를 저장하는 컬렉션이다. 따라서 중복 제거가 필요할 때 유용하게 사용할 수 있다.

## 셋 표현식
- 중괄호()를 사용하여 집합을 생성한다.

```python
safari_set = {"Bear", "Koala", "Panda", "Gorilla"}
numbers = {1, 2, 3, 4, 5}
mixed_set = {1, "Bear", (1, 2, 3)} # 튜플도 들어갈 수 있음
```

1. 순서가 없는 자료형이며, 순서가 없으므로 인덱스를 이용하여 접근 불가능

```python
ohgiraffers = {'Pig', 'Squirrel', 'Panda', 'Wolf'}
print(ohgiraffers)
# print(ohgiraffers[1]) # TypeError 발생
```

2. 중복을 허용하지 않으므로 동일한 값은 하나만 저장된다.

```python
ohgiraffers = {'Pig', 'Squirrel', 'Panda', 'Wolf', 'Panda'}
print(ohgiraffers)
```

3. 가변성을 가진 자료형으로, 데이터, 즉 요소를 추가하거나 제거할 수 있다.

```python
ohgiraffers.remove('Pig')
ohgiraffers.add("Gorilla")

print(ohgiraffers)
```

4. 수학적 집합 연산(합집합, 교집합, 차집합 등)을 지원한다.
5. in 키워드를 통해 포함 여부를 반환받을 수 있다.

```python
team = {'Gorilla', 'Tiger', 'Sheep', 'Panda'}
print('Panda' in team)
```

## 셋의 메서드
- 요소를 추가하는 메서드
1. add(값): set의 list에 인자로 전달받은 값을 추가한다.

```python
team.add('Elephant')
print(team)
```

2. update(값): 요소 값을 추가한다. 배열을 이용해 한 번에 여러 값을 추가할 수 있다.

```python
team.update(["Monkey", "Tiger", "Panda", "Gorilla"])
print(team)
```

3. remove(값): 특정 요소를 제거, 값이 존재하지 않으면 Error를 발생시킨다.

```python
# team.remove('Squirrel') # KeyError 발생
team.remove('Elephant')

print(team)
```

4. discard(값): 특정 요소를 제거, 값이 존재하지 않아도 Error가 발생하지 않는다.

```python
team.discard('Squirrel')
```

5. pop(): 임의의 값을 제거, 집합은 순서를 보장하지 않으므로 어떤 값이 제거될 지 예측할 수 없다.

```python
team.pop()
print(team)
```

6. clear(): 모든 값을 제거

```python
team.clear()
print(team)
```

## 집합 연산 메서드
1. union(): 두 set 자료형을 합친다 (합집합)

```python
java_team = {'Sheep', 'Tiger'}
python_team = {'Panda', 'Rabit', 'Koala', 'Tiger'}

ohgiraffers = java_team.union(python_team)
print(java_team | python_team)
print(ohgiraffers)
```

2. intersection(): 두 set 자료형의 교집합을 반환

```python
print(java_team.intersection(python_team))
print(java_team & python_team)
```

3. difference(): 두 set 자료형의 차집합을 반환한다. (좌항을 기준으로 우항의 차집합 반환)

```python
print(java_team.difference(python_team))
print(java_team - python_team)
```

4. symmetric_difference(): 대칭 차집합을 반환(양쪽 모두의 차집합을 반환한다.)

```python
print(java_team.symmetric_difference(python_team))
print(java_team ^ python_team)
```

## 복사 메서드
- copy(): 대상 set을 복사하여 반환한다.

```python
python_team = {'Panda', 'Rabit', 'Koala', 'Tiger'}
zzap_team = python_team.copy()

print(python_team)
print(zzap_team)

# copy()는 깊은 복사를 한다. (값만 복사한다.)
zzap_team.remove('Tiger')
print(python_team)
print(zzap_team)
```
