### 디큐 (deque)
- 파이썬의 collections 모듈에서 제공하는 자료구조로, 양쪽 끝에서 효율적인 삽입과 삭제가 가능하게 설계됨
#### 디큐의 메서드

```python
from collections import deque

dq = deque()

dq.append("a")
dq.append("b")
dq.append("c")

print(dq)
```

- append(), pop(): list의 append(), pop()과 동일

```python
# appendleft(): 왼쪽 끝에 요소를 추가
dq.appendleft('x')
print(dq)
# popleft(): 왼쪽 끝 요소 값 반환하며 제거
value = dq.pop()
print(value)

left_value = dq.popleft()
print(left_value)

print(dq)
```

### 튜플 (tuple)
#### 튜플이란
- 여러 개의 값을 하나의 데이터 구조로 묶어서 관리할 수 있는 불변(immutable) 시퀀스이다.

```python
# 튜플명 = ('값1', '값2', ...)
tuples = (1, 2, 'hello'), ('test', 1, 2, 3, 4)
print(tuples)
print(tuples[0])
print(tuples[0][2])

no_wrap_tuples = 1, 2, 3
print(no_wrap_tuples)
```

```python
# 불변 객체이므로 요소 값을 한 번 할당하면 추가, 수정, 삭제 불가능함
# tuples[0] = (1, 2, 3, 4)

testList = (1, 2, 3, 4)
print(id(testList))

testList = (3, 4, 5)
print(id(testList))
```

```python
# 튜플 간 연산

# 튜플 합치기
safari_tuple = ("Bear", "Gorilla", "Squarrel")
another_safari_tuple = ("Monkey", "Tiger", "Wolf")
print(safari_tuple + another_safari_tuple)

# 튜플 반복
print(safari_tuple * 2)

# 튜플의 길이
print(len(safari_tuple + another_safari_tuple))
```
