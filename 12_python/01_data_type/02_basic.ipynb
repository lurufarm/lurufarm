# 기본 자료형
## 숫자형(Numeric)

```python
# int
num1 = 1
num2 = -1

print(type(num1))
print(type(num2))

# float: 실수 값을 가지는 자료형
num3 = 3.14
num4 = 1.0

print(type(num3))
print(type(num4))
```

### 숫자 자료형의 연산
- 사칙연산(+, -, *, /)

```python
num1 = 10
num2 = 3

print(num1 + num2)
print(num1 - num2)
print(num1 * num2)
print(num1 / num2)

# 서로 다른 자료형의 피연산자를 갖는 경우, 정수형을 실수형으로 자동 변환
num3 = 10
num4 = 0.1

print(num3 + num4)
print(num3 / num4)
```

- 나눗셈 나머지 연산과 몫 연산

```python
# %: 나누기 연산을 하고 나머지를 반환한다.
num1 = 11
num2 = 7
print(num1 % num2)

# //: 몫만 구하는 연산
print(num1 // num2) 
```

- 제곱 연산

```python
base = 9
exponent = 2

print(base ** exponent)
print(exponent ** base)
```

### 논리형 (Boolean)
- 논리 자료형은 참 또는 거짓을 표현하는 자료형으로, True, False를 사용한다.
- 논리 자료형은 문자열이 아니므로 따옴표('',"")로 감싸서는 안되며, 첫 글자를 반드시 대문자로 해야만 한다.

```python
bool1 = True
bool2 = False

print(type(bool1))
print(type(bool2))
```

#### 논리 자료형의 연산
- 비교 연산: ==, !=, >, <, >=, <=

```python
print(1 == 3)
print('ohgiraffers' != 'Ohgiraffers')
```

- 논리 연산

```python
# and: 좌항과 우항의 비교 연산 결과 논리 자료형이 모두 True여야만 True 반환
print(1 != 3 and 100 > 99 and True)
print(1 != 3 and 100 > 99 and 100 < 99)

# or: 좌항과 우항의 비교 연산 결과 논리 자료형 중 하나라도 True이면 True 반환
print(1 != 3 or 100 > 99 or True)
print(1 != 3 or 100 > 99 or 100 < 99)
print(1 == 3 or False or 100 < 99)

# not: 우항에 있는 비교 연산 결과 또는 논리 자료형을 뒤집는다.
print(not 'ohgiraffers' == 'Ohgiraffers')
print(not True)
```

#### 문자형의 메서드
1. replace(): 문자열을 치환하는 메서드

```python
enroll_date = '2025/03/11'
rep_enroll_date = enroll_date.replace("/", ".")

print(rep_enroll_date)
```

2. strip(): 인자는 제거할 문자 집합을 지정하는 문자열이다. 인자를 생략하면 공백을 제거한다.

```python
origin = 'ohgiraffers'
with_white_space = '   oh giraffers   '

# 인자 생략 == 공백 제거
print(with_white_space.strip())
print(with_white_space)

# '   o' 까지 제거
print(with_white_space.strip('   o'))

# 'os' 제거
print(origin.strip('os'))

# lstrip(): 선행만 제거
print(origin.lstrip('os'))
# rstrip(): 후행만 제거
print(origin.rstrip('os'))
```

```python
# 대소문자 관련 메서드
origin_str = 'hELLO wORLD!'

print(origin_str.upper())           # 모두 대문자
print(origin_str.lower())           # 모두 소문자
print(origin_str.capitalize())      # 첫 글자만 대문자로 변경
print(origin_str.swapcase())        # 대문자->소문자, 소문자->대문자로 변경
print(origin_str.title())           # 단어의 첫 글자를 대문자로 변경
```

#### 문자형 포맷팅
1. % 포맷팅: 변수 포맷을 사용하여 문자열에 변수 값을 삽입할 수 있다. (오래된 방식)
- %s: 문자열
- %c: 문자
- %d: 정수
- %f: 실수

```python
x = 10
print("x is %d" %x)

y = "code"
print("y is %s" %y)
```

2. format(): 타입 명시 없이 format()함수를 이용할 수 있다.
- 단, 이 때 더 큰 범위로만 자동 형변환된다.

```python
x = 10
y = "code"
print("x is {0}".format(x))
print("x is {0} y is {1}".format(x, y))
```

3. f-string: 변수를 활용하여 문자열을 구성할 때, 따옴표 앞에 f를 붙이고, 따옴표 안에서 중괄호 안에 변수명을 넣어 활용할 수 있다.

```python
teacher_name = "판다"
print(f"안녕하세요. 오늘부터 여러분과 함께 공부할 {teacher_name} 강사입니다~~")
```

#### 형변환
1. 암시적 형변환

```python
print(True + 3)
print(3 + 5.0)
# print(3 + "5") # 암시적 형변환 불가능
```

2. 명시적 형변환

```python
# int(자료형): 문자형이나 숫자형 중 실수를 정수로 형변환하는 함수
print(int('3') + 4)
# print(int("3.5") + 2) # 문자열의 형태가 정수형인 경우에만 변경됨

# float(자료형): 문자형이나 숫자형 중 정수를 실수로 변환하는 함수
print(float('3'))

# str(자료형): 숫자형, 리스트, 튜플, 딕셔너리 등의 자료형을 문자로 형변환하는 함수
print(str(1))
print(str(1.0))
print(str({1, 2, 3})) # 집합

# list(자료형): 문자열, 튜플, 딕셔너리 등의 자료형을 리스트로 형변환하는 함수
print(list('ohgiraffers'))
print(list((1, 2, 3, 4, 5))) # 튜플 자료형
print(list({
    "1" : "사랑하는 우리반",
    "2" : "열심히 공부하십시오",
    "3" : "-판다쌤 왈"
    })) # 딕셔너리 자료형. key와 value로 이루어져있다.
```
