# 딕셔너리(Dictionaries)
## 딕셔너리란
- key와 value의 쌍으로 구성된 자료형으로, 키를 통해 값을 찾을 수 있다.

```python
# 딕셔너리명 = { 키1:값1, 키2:값2, ...}

teacher = { 'name':'bear', 'team':'ohgiraffers' }
print(type(teacher))

# key값을 활용한 value 탐색
print(teacher['name'])
# print(teacher.team)   # Error
```

- 키는 고유해야 하며 불변(immutable)한 데이터 타입이어야 한다.
- 만약 중복 키로 값을 입력하면, 기존 값을 덮어쓴다.

```python
ohgiraffers = {
    'Tiger' : 'JPA',
    'Bear' : 'Python',
    'Gorilla' : 'Spring',
    'Panda' : 'Java'
}

# invalid_key_teacher = {["name", "team"] : 'bear.ohgiraffers'} # TypeError. List로는 키값을 설정할 수 없다는 에러
valid_key_teacher = {("name", "team") : 'bear.ohgiraffers'} # TypeError나지 않음. Tuple로는 키값을 설정할 수 있음
print(valid_key_teacher)

ohgiraffers['Tiger'] = 'SQL'
print(ohgiraffers['Tiger'])

ohgiraffers['Koala'] = 'javascript'
print(ohgiraffers['Koala'])
print(ohgiraffers)

# del 키워드를 이용해 딕셔너리 삭제할 수 있다.

del ohgiraffers
# print(ohgiraffers) # 이미 삭제되어 Error뜸
```

```python
# get(): 매개변수로 전달받은 키에 해당하는 값 반환, 해당 키가 존재하지 않으면 None 반환
ohgiraffers = {
    'Tiger' : 'JPA',
    'Bear' : 'Python',
    'Gorilla' : 'Spring',
    'Panda' : 'Java'
}
print(ohgiraffers['Tiger'])
print(ohgiraffers.get('Tiger'))
print(ohgiraffers.get('test'))
```

```python
# keys(): 딕셔너리의 모든 키를 반환
print(ohgiraffers.keys())

# values(): 딕셔너리의 모든 값을 반환
print(ohgiraffers.values())

# items(): 딕셔너리의 모든 항목(키-값 쌍)을 반환
print(ohgiraffers.items())
```

```python
# pop(키): 매개변수의 값을 제거하고 반환한다.
print(ohgiraffers.pop('Tiger'))

# del 키워드를 이용해 특정키의 키-값 아이템을 제거할 수 있다.
del ohgiraffers['Bear']
print(ohgiraffers)
```

```python
# popitem(): 마지막 요소의 key-value 값을 제거하고 튜플 형태로 반환

poped_item = ohgiraffers.popitem()
print(poped_item)
print(ohgiraffers)
```

```python
# update(): 다른 딕셔너리나 키-값 쌍의 iterable을 사용하여 요소를 추가한다.
ohgiraffers.update({'test' : 'value'})
print(ohgiraffers)
```

```python
# clear(): 모든 항목을 제거한다.
ohgiraffers.clear()
print(ohgiraffers)
```
